```python
import pandas as pd

PARQUET_PATH = 'data/synthetic_csi_data_1440min.parquet'
OUTPUT_CSV   = 'data/csi_working.csv'

print("Reading parquet...")
df = pd.read_parquet(PARQUET_PATH, engine='fastparquet')
print(f"Full parquet shape: {df.shape}")

# Take first 1 million rows
df_cut = df.iloc[:1_000_000].reset_index(drop=True)

# Save to CSV
print(f"Saving to {OUTPUT_CSV}...")
df_cut.to_csv(OUTPUT_CSV, index=False)
print(f"Saved!")

```

```python
mpty  = df_cut[df_cut['occupancy'] == 0]
moving = df_cut[(df_cut['occupancy'] == 1) & (df_cut['movement'] == 1)]
still  = df_cut[(df_cut['occupancy'] == 1) & (df_cut['movement'] == 0)]

print(f"\n{'='*45}")
print(f"  Total rows            : {len(df_cut):>10,}")
print(f"  🔴 Empty room         : {len(empty):>10,}  ({len(empty)/len(df_cut)*100:.1f}%)")
print(f"  🟡 Occupied+Moving    : {len(moving):>10,}  ({len(moving)/len(df_cut)*100:.1f}%)")
print(f"  🟢 Occupied+Still     : {len(still):>10,}  ({len(still)/len(df_cut)*100:.1f}%)")
print(f"  Heart rate range      : {still['heart_rate_bpm'].min():.1f} – {still['heart_rate_bpm'].max():.1f} BPM")
print(f"  Complete 10s windows  : {len(still) // 2000:>10,}")
print(f"{'='*45}")
```
